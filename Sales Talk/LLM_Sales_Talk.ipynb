```python
from openai import OpenAI
from getpass import getpass
import pandas as pd

# API 키 설정
api_key = getpass('OpenAI API 키를 입력하세요: ')
client = OpenAI(api_key=api_key)

# GPT-4o-mini와 대화하는 함수
def chat_with_gpt4omini(prompt, max_tokens=200):
    try:
        response = client.chat.completions.create(
            model="gpt-4o-mini",
            messages=[
                {"role": "system", "content": "너는 iptv를 추천하는 sk 브로드밴드 세일즈 맨이야. 인터넷을 사용하는 고객 특성을 사용해서 iptv를 가입할 수 있게끔 하는 게 목적이야"},
                {"role": "user", "content": prompt}
            ],
            max_tokens=max_tokens,
            temperature=0.2
        )
        return response.choices[0].message.content.strip()
    except Exception as e:
        return f"An error occurred: {str(e)}"

# 고객 특성 → 자연스러운 추천 문장 템플릿 매핑
mean1_to_template = {
    "기가인터넷 사용": "고속 인터넷을 사용 중이니 IPTV 이용에 적합합니다.",
    "데이터 사용량 많음": "인터넷 사용량이 많은 편이라 IPTV 이용에 적합합니다.",
    "와이파이 사용": "무선 환경에서 IPTV도 손쉽게 이용하실 수 있어요.",
    "윙즈(와이파이 확장기) 사용": "SK 브로드밴드의 윙즈는 와이파이 신호를 확장하여 IPTV와 같은 서비스도 원활하게 이용할 수 있게 도와줍니다.",
    "장기고객": "장기간 이용 고객님께는 IPTV 추가 시 혜택도 더해드립니다.",
    "재약정함": "재약정을 하셨다면 IPTV 추가 시 추가 혜택을 적용받을 수 있습니다.",
    "와이파이 접속 기기 대수 많음": "인터넷 사용량이 많은 편이라 IPTV 이용에 적합합니다.",
    "데이터 송신량 많음": "인터넷 사용량이 많은 편이라 IPTV 이용에 적합합니다.",
    "데이터 수신량 많음": "인터넷 사용량이 많은 편이라 IPTV 이용에 적합합니다."
}

# 무한 루프 (고객 특성 입력 및 Sales Talk 생성)
while True:
    # 고객 특성 입력 받기 (쉼표로 구분하여 입력)
    customer_input = input("고객 특성을 입력하세요 (예: 데이터 사용량 많음, 기가인터넷 사용) 또는 '끝'을 입력하여 종료: ")

    # "끝"을 입력하면 종료
    if customer_input.lower() == "끝":
        print("프로그램을 종료합니다.")
        break

    # 고객 특성을 쉼표로 분리
    input_list = [x.strip() for x in customer_input.split(",")]

    # 고객 특성에 맞는 추천 이유들 결합
    combined_mean1 = "\n- ".join(input_list)
    combined_mean2 = "\n- ".join([mean1_to_template.get(mean1, "적합한 추천 이유를 찾을 수 없습니다.") for mean1 in input_list])

    # Sales Talk를 하나로 생성
    prompt = f"""
고객의 특성은 다음과 같습니다:{combined_mean1}

추천 이유는 다음과 같습니다:{combined_mean2}

이 정보를 기반으로 IPTV 가입을 설득력 있게 유도하는 자연스러운 세일즈 토크 문장을 2문장 이내로 작성해줘.
친절하지만 전문가답고, 과하게 영업티는 나지 않게 해줘.

[조건]
- 친절하지만 전문가답고, 과하게 영업티는 나지 않게 해줘.
- 반드시 추천 근거만 활용하고, 가족, 여가, 취향 등 추가적인 부가 설명은 넣지 마.
- 입력된 특성을 반복해서 언급하지 말고, 중복된 내용은 한 번만 언급해.
- 문장은 간결하게, 2문장 이내로 작성해.
- 반드시 고객의 특성과 추천 이유 내에 명시된 내용만 활용해 문장을 작성할 것
- 입력한 고객 특성을 모두 사용해 줘
"""

    # Sales Talk 생성
    sales_talk = chat_with_gpt4omini(prompt)

    # 결과 출력
    print("\nGenerated Sales Talk:")
    print(sales_talk)

    # 결과를 데이터프레임에 저장
    result_df = pd.DataFrame({
        "mean1": [combined_mean1],
        "sales_talk": [sales_talk]
    })

    # 결과를 엑셀로 저장
    result_df.to_excel("최종_세일즈톡결과.xlsx", index=False)
    #print("결과가 엑셀 파일로 저장되었습니다.")
    print("="*100)
```
