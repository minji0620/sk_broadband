```python
import pandas as pd
from google.colab import files

# CSV 파일 경로
file_path = "shap_분석결과.csv"

# 데이터프레임으로 읽어오기
df = pd.read_csv(file_path)

# mean1 컬럼에서 변경할 값들
mean1_mapping = {
    "와이파이 접속 기기 대수 많음": "데이터 사용량 많음",
    "데이터 수신량 많음": "데이터 사용량 많음",
    "데이터 송신량 많음": "데이터 사용량 많음"
}

# mean1 컬럼 값 변경
df['mean1'] = df['mean1'].replace(mean1_mapping)

# mean2 컬럼에서 변경할 값
mean2_mapping = {
    "인터넷 사용량과 콘텐츠 이용량이 높은 편에 속함": "인터넷 사용량이 많은 편이라 IPTV 이용에 적합합니다."
}

# mean2 컬럼 값 변경
df['mean2'] = df['mean2'].replace(mean2_mapping)

# 수정된 데이터프레임을 새로운 CSV 파일로 저장 (덮어쓰기)
df.to_csv("modified_" + file_path, index=False, encoding='utf-8')

# 수정된 파일 다운로드
files.download("modified_" + file_path)

print("CSV 파일이 수정되었고 다운로드됩니다.")
```
